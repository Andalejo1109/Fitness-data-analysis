```python
from garminconnect import Garmin
import datetime
import pandas as pd
import matplotlib.pyplot as plt
```

```python
username= "alejo1109@gmail.com"
password= "AArL110989"
```

```python
#Initialize the Garmin API Client
api= Garmin(username, password)

#attemt to login
try:
    api.login()
    print("login successful")
except Exception as e:
    print(f"An error occured while login in:{e}")

#set the start and end date nine months apart
activity_start_date = datetime.date(2022, 10, 12)
activity_end_date = datetime.date(2024, 3,12)

#after successful logint attempt to fetch activities

try:
    activities = api.get_activities_by_date(activity_start_date.isoformat(), activity_end_date.isoformat())
    print(f"Fetched {len(activities)} activities.")
except Exception as e:
    print(F"An error occurred while fetching activities: {e}")
    
```

```python
activities_df= pd.DataFrame(activities)
```
